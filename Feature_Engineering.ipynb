## Date Time Extraction

## Cyclical Time:
    When working with time seris data, you in some cases help the model to understand the cyclical nature of time. The model on its own, perceive the maximu value, to be very far away from the minimum value("which in untrue"), so we help the model understand the cyclic nature by transforming data into a format where the minimum value, becomes the immediate next value to the maximum. 
    This is done using the Sine and Cosine transformation.
    

```python
data_T = pd.DataFrame()
data_T["DateTime"] = pd.date_range(start = "2025-01-01", periods = 24, freq = 'H')
data_T
```

#### Separate clock from calendar

```python
data_T['Hour'] = data_T['DateTime'].dt.hour
data_T.head()
```

#### Perform transformation

##### Hour cyclical transformation

```python
#USing numpy cos and sine fuunction to perform transformation
import numpy as np
data_T['Hour_Sin'] = np.sin(2 * np.pi * data_T['Hour']   / 24 )
data_T['Hour_Cos'] = np.cos(2 * np.pi * data_T['Hour']   / 24 )
```

```python
data_T.head()
```

```python
plt.figure(figsize = (6,6))
plt.plot(data_T['Hour_Cos'], data_T['Hour_Sin'], marker = 'o')
for i in  range(24):
    plt.text(data_T['Hour_Cos'][i], data_T['Hour_Sin'][i], str(data_T['Hour'][i]))
plt.show()
```

## Aggregation
    In analysis, aggregation means combining or summarizing data to make it easier to interpret or draw insights from.
this is done using the pandas "groupby()" function. 
#### Groupby()
    This is a powerful pandas function used in performing aggregation,summation and other statistical and mathematical functions, or any other computation, where you have to combine two features.

```python
data.groupby(["Brand", "Registration"]).count()
```

```python
data.groupby(["Brand", "Registration"])["Engine Type"].count()
```

```python
data.groupby("Brand").agg({"Price": "sum"})
```

```python
data.groupby("Brand").agg({"Price": "sum"}).reset_index().sort_values(by="Price",ascending=False)   #to arrrange it in ascending order
```
